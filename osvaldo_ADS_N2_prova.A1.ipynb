# 1 Apresente em tela (output) toda a base de dados. Pontuação: 0,5 ponto


```python
# Apresentando meu DataFrame sobre animes!

import pandas as pd

nomes_animes=["My Hero Academy", "One Piece", "Naruto", "Bleach", "Vinland Saga", "Dragon Ball"]
autores=["Kohei Horikoshi", "Eiichiro Oda", "Masashi Kishimoto", "Tite Kubo", "Makoto Yukimura", "Akira Toriyama"]
protagonista=["Midoria Izuku", "Monkey D. Luffy", "Naruto Uzumaki", "Ichigo", "Torfinn Karself", "Goku"]
viloes=["All For One", "Marinha", "Ootsutsuki", "Yhwach", "Askeladd", "Deuses da Guerra"]
musicas=["Peace Sing", "The Peak", "Blue Bird", "All Song Are Here", "Troll", "Sorriso Resplandescente"]
ano_de_lancamento=[2016, 1997, 1999, 2004, 2005, 1986]

dados_animes={
    "Nomes de Animes": nomes_animes,
    "Autores": autores,
    "Protagonistas": protagonista,
    "Vilões": viloes,
    "Musicas": musicas,
    "Ano de Lançamento": ano_de_lancamento
}

dados_animes=pd.DataFrame(dados_animes)

display(dados_animes)
```

# 2 Apresente o tamanho do seu dataframe (quantas colunas x linhas). Pontuação: 0,5 pontos

```python
# Sabendo o tamanho do meu DataFrame!

# shape : retorna a dimensão do DataFrame (número de linhas e colunas).

display("O tamanho de linhas e colunas é:")
display(dados_animes.shape)
```

# 3 Acesse a linha (x) e apresente em tela todas as características do item. Pontuação: 0,5 pontos


```python
# Verificando a linha acessada!

#sample()é usado para gerar uma linha ou coluna aleatória de amostra do quadro de dados do chamador da função.

display("A linha apresentada foi a:")
dados_animes.sample(1)
```

# 4 Verifique se o dataframe está vazio. Pontuação: 0,5 pontos

```python
# Verificando se meu DataFrame está vazio!

# empty: retorna se o DataFrame está vazio (True) ou não (False).

display("Data frame está vazio?")
dados_animes.empty
```

# 5 Apresente em tela os 5 primeiros registros da base de dados. Pontuação: 0,5 pontos


```python
# Mostrando os 5 primeiros registros da minha base de dados.

# head (n) retorna as primeiras n linhas de um conjuto de dados. Como padrão, df. head() retorna as 5 primeiras linhas do Data

display("Os 5 primeiros números apresentados são:")
dados_animes.head(5)
```

# 6 Exclua um item (linha) de sua base de dados. Pontuação: 0,5 pontos



```python
# Excluindo uma linha do meu DataFrame.

# drop() permite remover linhas ou colunas de um DataFrame com base em rótulos ou índices.
# Para excluir uma coluna, basta especificar o rótulo da coluna e o parâmetro axis=1.
# O parâmetro axis indica o que deve ser removido, sendo 0 para linhas e 1 para colunas.

dados_animes.drop(index = [1], inplace = True)
display(dados_animes)0,
```

# 7 Adicione um item (linha) na sua base de dados. Pontuação: 0,5 pontos




```python
# Adicionando uma linha na minha base de dados.

#loc é utilizada para selecionar e filtrar dados em um DataFrame com base em rótulos de
# linhas e colunas, oferecendo uma maneira eficaz de acessar informações específicas de um conjunto de dados.

dados_animes.loc[6] = ["Solo Leving", "Chu-Gong", "Sung-Jinwoo", "Monarcas", "LEveL", "2024"]
display(dados_animes)
```


# 8 Transponha a coluna para a linha em sua base de dados. Pontuação: 0,5 pontos



```python
# Transpondo a coluna para a base de dados.

# transpose()função Pandas transpõe o índice e as colunas do dataframe.
# Ele reflete o DataFrame sobre sua diagonal principal, gravando linhas como colunas e vice-versa.

dados_animes = dados_animes.transpose()

display(dados_animes)
```

# 9 Apresente em tela somente a 1ª e a 2ª coluna (rótulo) da base de dados. Pontuação: 0,5 pontos



```python
# Apresentando a tela da coluna 1 e 2!

# get()função Pandas é usada para obter um item do objeto para uma determinada
# tecla. A chave pode ser uma ou mais colunas de dataframe. Ele retorna o valor padrão se não for encontrado.

dados_animes.get([0,2])
```
